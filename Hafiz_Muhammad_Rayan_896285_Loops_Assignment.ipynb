## Write a program to count how many digits exist in a number using a loop.
### Example:

* Input: 45892
* Output: 5

```python
number = abs(int(input("Input: ")))   # abs() so negative numbers also work

digit_count = 0

if number == 0:
    digit_count = 1                   # 0 has exactly one digit
else:
    while number > 0:
        number = number // 10         # remove the last digit
        digit_count += 1              # count the removed digit

print("Output:", digit_count)
```

## Write a program to calculate the factorial of a number.
### Example:

* Input : 5
* Output 120

```python
number = int(input("Input: "))
if number < 0:
    print("Factorial is not defined for negative numbers.")
else:
    factorial = 1                     # 0! = 1, and 1 is the starting value for multiplying
    for value in range(1, number + 1):
        factorial = factorial * value
    print("Output:", factorial)
```

## Student Marks Entry

### Goal:
    Record student marks and skip invalid entries.

### Steps:

    Create an empty list marks = [].
    
    Ask user to enter marks for 5 students.
    
    If marks < 0 or > 100 → print “Invalid marks” and continue.
    
    If user enters -1 → stop taking input (break).

### Example:

    Enter marks: 90  
    Enter marks: 105 → Invalid marks  
    Enter marks: 80  
    Enter marks: -1 → stop

```python
marks = []
for student in range(5):               # marks for 5 students
    entered_marks = int(input("Enter marks: "))
    if entered_marks == -1:            # must be checked before the invalid check
        break                          # -1 means stop taking input
    if entered_marks < 0 or entered_marks > 100:
        print("Invalid marks")
        continue                       # skip this entry
    marks.append(entered_marks)
print("Valid Marks:", marks)
if len(marks) > 0:
    average = sum(marks) / len(marks)
    print("Average =", average)
else:
    print("No valid marks were entered.")
```

## Course Registration System

### Goal:
    Allow students to register for available courses.

### Steps:

    courses = ["AI", "Python", "Math", "Physics"]
    
    Loop for multiple registrations.
    
    If course not in list → print “Not available” and continue
    
    If user enters “exit” → break
    
    Show registered courses.

### Example:

    Enter course: Python  
    Enter course: Biology → Not available  
    Enter course: AI  
    Enter course: exit  
    Registered: ['Python', 'AI']


```python
courses = ["AI", "Python", "Math", "Physics"]
registered = []
while True:
    course = input("Enter course: ").strip()
    if course.lower() == "exit":
        break
    # match the course name ignoring upper/lower case
    matched_course = ""
    for available_course in courses:
        if available_course.lower() == course.lower():
            matched_course = available_course
    if matched_course == "":
        print("Not available")
        continue
    if matched_course in registered:
        print("Already registered")
        continue
    registered.append(matched_course)

print("Registered:", registered)
```

## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
account_created = False
account_holder = ""
balance = 0.0
while True:
    print()
    print("===== Banking System =====")
    print("1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Check Balance")
    print("5. Exit")
    choice = input("Enter your choice (1-5): ")
    if choice == "1":
        if account_created:
            print("An account already exists for", account_holder)
        else:
            account_holder = input("Enter account holder name: ")
            opening_deposit = float(input("Enter opening deposit: "))
            if opening_deposit < 0:
                print("Opening deposit cannot be negative. Account not created.")
            else:
                balance = opening_deposit
                account_created = True
                print("Account created for", account_holder)

    elif choice == "2":
        if not account_created:
            print("Please create an account first.")
        else:
            deposit_amount = float(input("Enter amount to deposit: "))
            if deposit_amount <= 0:
                print("Deposit must be greater than 0.")
            else:
                balance += deposit_amount
                print("Deposited successfully. New balance:", format(balance, ".2f"))

    elif choice == "3":
        if not account_created:
            print("Please create an account first.")
        else:
            withdraw_amount = float(input("Enter amount to withdraw: "))
            if withdraw_amount <= 0:
                print("Withdrawal must be greater than 0.")
            elif withdraw_amount > balance:
                print("Insufficient balance. Current balance:", format(balance, ".2f"))
            else:
                balance -= withdraw_amount
                print("Withdrawn successfully. New balance:", format(balance, ".2f"))

    elif choice == "4":
        if not account_created:
            print("Please create an account first.")
        else:
            print("Account holder:", account_holder)
            print("Current balance:", format(balance, ".2f"))

    elif choice == "5":
        print("Thank you for banking with us. Goodbye!")
        break

    else:
        print("Invalid choice. Please enter a number from 1 to 5.")
```

## School Marks Report
### Goal:

    Take input for multiple students’ marks in 3 subjects and calculate their total and average.

### Steps:

    Outer loop → for each student (3 students).
    
    Inner loop → take marks for 3 subjects.
    
    Store each student’s total in a list.
    
    Print all totals and class average.

### Example:
    Enter marks for Student 1:
    Subject 1: 80
    Subject 2: 75
    Subject 3: 90
    Enter marks for Student 2:
    Subject 1: 60
    Subject 2: 70
    Subject 3: 65
    → Output: [245, 195]

```python
student_totals = []
for student in range(1, 4):                       # 3 students
    print("Enter marks for Student " + str(student) + ":")
    student_total = 0
    for subject in range(1, 4):                   # 3 subjects
        subject_marks = int(input("Subject " + str(subject) + ": "))
        student_total = student_total + subject_marks
    student_totals.append(student_total)
print("Output:", student_totals)
for index in range(len(student_totals)):
    student_average = student_totals[index] / 3
    print("Student", index + 1, "-> Total =", student_totals[index], ", Average =", round(student_average, 2))
# class average = average of all the students' totals
class_average = sum(student_totals) / len(student_totals)
print("Class Average =", round(class_average, 2))
```

## Shop Sales Tracker
### Goal:

    Track daily sales of 3 products over 5 days and calculate the total revenue.

### Steps:

    Outer loop → iterate over 5 days.
    
    Inner loop → take sales amount for 3 products each day.
    
    Store daily total in a list.
    
    If any product sale = 0 → continue (skip that product).

    Product 2: 150
    Product 3: 0 → skipped

## Student Attendance Tracker
### Goal:

    Record attendance (Present/Absent) for multiple classes and students.

### Steps:

    Outer loop → each class (2 classes).
    
    Inner loop → each student (3 students).
    
    Store "P" or "A" in a nested list.

    Print full attendance chart.

### Example:
    Class 1:
    Student 1 (P/A): P
    Student 2 (P/A): A
    Student 3 (P/A): P
    Class 2:
    Student 1 (P/A): P
    Student 2 (P/A): P
    Student 3 (P/A): P
    Output: [['P', 'A', 'P'], ['P', 'P', 'P']]


```python
attendance = []                                   # nested list: one list per class
for class_number in range(1, 3):                  # 2 classes
    print("Class " + str(class_number) + ":")
    class_attendance = []
    for student in range(1, 4):                   # 3 students
        status = input("Student " + str(student) + " (P/A): ").strip().upper()
        # keep asking until the user types P or A
        while status != "P" and status != "A":
            print("Please enter P or A only.")
            status = input("Student " + str(student) + " (P/A): ").strip().upper()
        class_attendance.append(status)
    attendance.append(class_attendance)
print("Output:", attendance)

print()
print("Attendance chart:")
for class_index in range(len(attendance)):
    print("Class", class_index + 1, ":", attendance[class_index])
```

## Cinema Seat Booking System
### Goal:

    Take user input for seat availability across rows and columns, and mark booked seats.

### Steps:

    Outer loop → for each row (3 rows).
    
    Inner loop → for 5 seats in that row.
    
    Ask if seat is booked (Y/N).
    
    If booked → add “Booked” else “Available”.
    
    Store in nested list and print seating chart.

### Example:
    Row 1:
    Seat 1 (Y/N): Y
    Seat 2 (Y/N): N
    Seat 3 (Y/N): Y
    Seat 4 (Y/N): N
    Seat 5 (Y/N): N
    Row 2:
    Seat 1 (Y/N): Y
    Seat 2 (Y/N): Y
    Seat 3 (Y/N): Y
    Seat 4 (Y/N): Y
    Seat 5 (Y/N): N
    
    Output:
    [['Booked','Available','Booked','Available','Available'],
     ['Booked','Booked','Booked','Booked','Available']]

```python
seating_chart = []                                # nested list: one list per row
for row in range(1, 4):                           # 3 rows
    print("Row " + str(row) + ":")
    row_seats = []
    for seat in range(1, 6):                      # 5 seats per row
        booked = input("Seat " + str(seat) + " (Y/N): ").strip().upper()
        # keep asking until the user types Y or N
        while booked != "Y" and booked != "N":
            print("Please enter Y or N only.")
            booked = input("Seat " + str(seat) + " (Y/N): ").strip().upper()
        if booked == "Y":
            row_seats.append("Booked")
        else:
            row_seats.append("Available")
    seating_chart.append(row_seats)
print()
print("Seating chart:")
for row_index in range(len(seating_chart)):
    print("Row", row_index + 1, ":", seating_chart[row_index])
```

## Multiplication Table Generator
### Goal:

    Generate multiplication tables for multiple numbers using nested loops.

### Steps:

    Outer loop → for numbers (2 to 5).
    
    Inner loop → multiply each number by 1–10.
    
    Print table for each number.
    
    Skip number 4 using continue.

### Example:
    2 × 1 = 2  
    ...  
    2 × 10 = 20  
    3 × 1 = 3  
    ...  
    (skip 4)  
    5 × 1 = 5  
    ...

## Hotel Room Booking Tracker
### Goal:

    Record bookings for rooms on each floor.

### Steps:

    Outer loop → for 3 floors.
    
    Inner loop → for 4 rooms per floor.
    
    Ask user if booked (Y/N).
    
    Count total booked and available rooms.
    
    Print summary per floor.

### Example:
    Floor 1:
    Room 1: Y
    Room 2: N
    Room 3: N
    Room 4: Y
    Floor 2:
    Room 1: Y
    Room 2: Y
    Room 3: Y
    Room 4: Y
    → Output: Floor 1 = 2 booked, Floor 2 = 4 booked

```python
total_booked = 0
total_available = 0
for floor in range(1, 4):                         # 3 floors
    print("Floor " + str(floor) + ":")
    floor_booked = 0
    floor_available = 0
    for room in range(1, 5):                      # 4 rooms per floor
        booked = input("Room " + str(room) + " (Y/N): ").strip().upper()
        # keep asking until the user types Y or N
        while booked != "Y" and booked != "N":
            print("Please enter Y or N only.")
            booked = input("Room " + str(room) + " (Y/N): ").strip().upper()
        if booked == "Y":
            floor_booked += 1
        else:
            floor_available += 1
    total_booked += floor_booked
    total_available += floor_available
    print("Floor", floor, "=", floor_booked, "booked,", floor_available, "available")
print()
print("Total booked rooms:", total_booked)
print("Total available rooms:", total_available)
```

## Invoice Tax Calculator
### Goal:

Take multiple items in multiple invoices and calculate tax per invoice.

### Steps:

Outer loop → for 3 invoices.

Inner loop → 4 items each.

Input item price, add to list.

Calculate total + 10% tax.

Print invoice total.

### Example:
    Invoice 1:
    Items: [100, 200, 300, 400]
    Invoice 2:
    Items: [250, 150, 350, 250]
    → Totals: [1100, 1100]

```python
invoice_totals = []
for invoice in range(1, 4):                       # 3 invoices
    print("Invoice " + str(invoice) + ":")
    item_prices = []
    for item in range(1, 5):                      # 4 items per invoice
        price = int(input("Item " + str(item) + " price: "))
        item_prices.append(price)
    subtotal = sum(item_prices)
    tax = subtotal * 0.10                         # 10% tax
    invoice_total = round(subtotal + tax, 2)
    print("Items:", item_prices)
    print("Subtotal =", subtotal, ", Tax =", round(tax, 2), ", Total =", invoice_total)
    invoice_totals.append(invoice_total)
print("Totals:", invoice_totals)
```

## Student Result Sheet
### Goal:

    Generate report cards for each student in a class.

### Steps:

    Outer loop → 5 students.
    
    Inner loop → 4 subjects.

    If marks < 33 → print “Failed in Subject”.

    Calculate total and grade per student.

### Example:
    Student 1: [80, 90, 70, 85] → Grade: A  
    Student 2: [50, 32, 45, 40] → Failed in Subject 2

```python
# Grade scale (average marks): A >= 80, B >= 70, C >= 60, D >= 50, E >= 33
results = []                                      # one result line per student
for student in range(1, 6):                       # 5 students
    print("Student " + str(student) + ":")
    student_marks = []
    failed_subjects = []
    for subject in range(1, 5):                   # 4 subjects
        marks = int(input("Subject " + str(subject) + " marks: "))
        student_marks.append(marks)
        if marks < 33:
            print("Failed in Subject", subject)
            failed_subjects.append(subject)
    total = sum(student_marks)
    average = total / 4
    if len(failed_subjects) > 0:
        subject_text = ""
        for subject in failed_subjects:
            if subject_text != "":
                subject_text = subject_text + ", "
            subject_text = subject_text + str(subject)
        result_line = "Student " + str(student) + ": " + str(student_marks) + " -> Total = " + str(total) + ", Failed in Subject " + subject_text
    else:
        if average >= 80:
            grade = "A"
        elif average >= 70:
            grade = "B"
        elif average >= 60:
            grade = "C"
        elif average >= 50:
            grade = "D"
        else:
            grade = "E"
        result_line = "Student " + str(student) + ": " + str(student_marks) + " -> Total = " + str(total) + ", Grade: " + grade
    results.append(result_line)
print()
print("===== RESULT SHEET =====")
for line in results:
    print(line)
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
features = ["Hotel", "Transport", "Food"]
for package in range(1, 4):                       # 3 packages
    print("Package " + str(package) + ":")
    ratings = []
    rating_sum = 0
    rating_count = 0
    for feature in features:                      # 3 features
        rating = int(input(feature + " rating (0-5): "))
        if rating < 0 or rating > 5:
            print("Invalid rating, skipped")
            continue
        ratings.append(rating)
        if rating == 0:
            continue                              # a rating of 0 is skipped in the average
        rating_sum = rating_sum + rating
        rating_count = rating_count + 1
    if rating_count > 0:
        average_rating = rating_sum / rating_count
        print("Package", package, ":", ratings, "-> Avg =", round(average_rating, 2))
    else:
        print("Package", package, ":", ratings, "-> No ratings to average")
```
